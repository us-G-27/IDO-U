# Práctica de Análisis de Regresión: Modelo MCO Simple

**Materia:** Investigación de Operaciones / Métodos Estadísticos  
**Objetivo:** Estimar los parámetros $\beta_0$ y $\beta_1$ mediante Mínimos Cuadrados Ordinarios (MCO) y evaluar la bondad de ajuste ($R^2$).

### Formulación del Modelo
$$Y_i = \beta_0 + \beta_1 X_i + \varepsilon_i$$

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Datos de ejemplo
X = np.array([2, 4, 6, 8, 10, 12, 14, 16], dtype=float)
Y = np.array([5.1, 9.2, 13.5, 17.8, 20.9, 25.1, 29.3, 33.0], dtype=float)

# 2. Estimación de MCO
x_bar, y_bar = np.mean(X), np.mean(Y)
beta_1 = np.sum((X - x_bar) * (Y - y_bar)) / np.sum((X - x_bar)**2)
beta_0 = y_bar - beta_1 * x_bar

# 3. Métricas
y_hat = beta_0 + beta_1 * X
residuos = Y - y_hat
r2 = 1 - (np.sum(residuos**2) / np.sum((Y - y_bar)**2))

print(f"Modelo estimado: Y_hat = {beta_0:.3f} + {beta_1:.3f} * X")
print(f"Coeficiente R²: {r2:.4f}")
```

```python
plt.figure(figsize=(8, 4.5))
plt.scatter(X, Y, color='#1f77b4', s=60, label='Observaciones reales')
plt.plot(X, y_hat, color='#d62728', linewidth=2, label=f'Recta MCO: $\hat{{Y}} = {beta_0:.2f} + {beta_1:.2f}X$')

plt.title('Ajuste de Regresión Lineal Simple por MCO', fontsize=12)
plt.xlabel('Variable Explicativa (X)', fontsize=10)
plt.ylabel('Variable Respuesta (Y)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
